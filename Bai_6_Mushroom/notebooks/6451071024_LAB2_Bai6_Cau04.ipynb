### Yêu cầu

Tách các thuộc tính đầu vào và thuộc tính quyết định.

### Dữ liệu và phương pháp

Thuộc tính quyết định: class. Sử dụng toàn bộ dữ liệu được cung cấp. Chia huấn luyện 70%, kiểm thử 30%, random_state=42 theo đề.

```python
print("Đặng Ngọc Hiếu - 6451071024")

from pathlib import Path
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
while not (ROOT / "datasets").is_dir():
    if (ROOT / "6451071024_Lab2" / "datasets").is_dir():
        ROOT = ROOT / "6451071024_Lab2"
        break
    if ROOT == ROOT.parent:
        raise FileNotFoundError("Không tìm thấy thư mục datasets của LAB2.")
    ROOT = ROOT.parent
pd.set_option("display.max_columns", None)


raw = pd.read_csv(ROOT / "datasets/mushroom/mushrooms.csv")
TARGET = "class"
if TARGET not in raw or raw.empty or raw.isna().any().any():
    raise ValueError("Kiểm tra Mushroom: cần cột class và các ô dữ liệu được cung cấp.")
if not set(raw[TARGET].unique()).issubset({"p", "e"}):
    raise ValueError("Nhãn Mushroom phải là p hoặc e.")
# Giữ ? là mức chưa xác định của stalk-root; đề bài này không yêu cầu xóa dòng.
data = raw.copy()
```

```python
print("Đặng Ngọc Hiếu - 6451071024")

features = data.drop(columns=TARGET)
labels = data[TARGET].copy()
if labels.isna().any() or features.empty:
    raise ValueError("Dữ liệu cần có đặc trưng và nhãn không trống.")
print("Thuộc tính quyết định:", TARGET)
print("Kích thước features / labels:", features.shape, labels.shape)
display(features.head(), labels.head().to_frame())
```
