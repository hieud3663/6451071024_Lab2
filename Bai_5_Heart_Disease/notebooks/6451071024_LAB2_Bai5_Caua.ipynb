# Heart Disease

## Câu a. Làm sạch và đổi tên thuộc tính

### Yêu cầu

Nhập heart.csv, xóa những dòng dữ liệu trống và đổi đủ 14 tên thuộc tính theo yêu cầu a trang 26.

### Dữ liệu và phương pháp

Thuộc tính quyết định: target. Sử dụng toàn bộ dữ liệu được cung cấp.

```python
print("Đặng Ngọc Hiếu - 6451071024")

from pathlib import Path
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
while not (ROOT / "datasets").is_dir():
    if (ROOT / "6451071024_Lab2" / "datasets").is_dir():
        ROOT = ROOT / "6451071024_Lab2"
        break
    if ROOT == ROOT.parent:
        raise FileNotFoundError("Không tìm thấy thư mục datasets của LAB2.")
    ROOT = ROOT.parent
pd.set_option("display.max_columns", None)


raw = pd.read_csv(ROOT / "datasets/heart/heart.csv", skipinitialspace=True)
if raw.shape[1] != 14 or raw.empty:
    raise ValueError("Dữ liệu Heart phải có đủ 14 cột.")
print("Tên cột trước đổi:", list(raw.columns))
print("Số dòng trước làm sạch:", len(raw))
display(raw.head())
data = raw.dropna().rename(columns={
    "cp": "chest_pain_type",
    "trestbps": "resting_blood_pressure",
    "chol": "cholesterol",
    "fbs": "fasting_blood_sugar",
    "restecg": "rest_ecg",
    "thalach": "max_heart_rate_achieved",
    "exang": "exercise_induced_angina",
    "oldpeak": "st_depression",
    "slope": "st_slope",
    "ca": "num_major_vessels",
    "thal": "thalassemia",
}, errors="raise")
TARGET = "target"
if not set(data[TARGET].unique()).issubset({0, 1}):
    raise ValueError("Nhãn Heart chỉ nhận 0 hoặc 1.")
print("Tên cột sau đổi:", list(data.columns))
print("Số dòng sau làm sạch:", len(data))
display(data.head())
```
